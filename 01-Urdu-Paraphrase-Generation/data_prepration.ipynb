```
import pandas as pd
df = pd.read_excel (r'/content/drive/My Drive/PPDB lexical training/test_data/Copy-of-lexical_test.xlsx', names=['a','b','c','d','e','f','g'])
```

```
df=df[:11000]
df
```

```
df['c']=df['c'].fillna(df['b'])
df['f']=df['f'].fillna(df['e'])
```

```
data=df[['c','f']]
data["c"] = data['c'].str.replace('[^\w\s]','')
data["c"] = data['c'].str.replace('\d+', '')
data["f"] = data['f'].str.replace('[^\w\s]','')
data["f"] = data['f'].str.replace('\d+', '')
data["c"] = data['c'].replace("\t","", regex=True)
data["c"] = data['c'].replace("\n","", regex=True)
data["f"] = data['f'].replace("\t","", regex=True)
data["f"] = data['f'].replace("\n","", regex=True)
```

```
data = data.dropna()
```

```
data = data[~data['c'].str.contains("[a-zA-Z]").fillna(False)]
data = data[~data['f'].str.contains("[a-zA-Z]").fillna(False)]
data = data[~data['c'].str.contains(r'\d')]
data = data[~data['f'].str.contains(r'\d')]
```

```
data
```

```
data=data[:10500]
```

```
data.to_csv('/content/drive/My Drive/PPDB lexical training/test_data/lexical_test_data.txt', sep='\t',header=False,index=False)
data.to_excel('/content/drive/My Drive/PPDB lexical training/test_data/lexical_test_data.xlsx',header=False,index=False)
```
